## Importing the libraries

```python
import pandas as pd
import numpy as np
```

## Importing the dataset

```python
dataset = pd.read_csv('train.csv')
print(dataset.head())
```

```python
# X = dataset.drop('Survived', axis=1).values
X = dataset[['Pclass', 'Sex', 'Age', 'SibSp',
             'Parch', 'Fare', 'Embarked']].values
y = dataset['Survived'].values
```

```python
print(X)
```

```python
print(y)
```

## Check missing data

```python
dataset.isnull().sum()
```

## Taking care of Missing data

```python
print(X[:,2])  ## checking missing values for AGE 
```

### Age: Age is numeric data, so we apply mean strategy

```python
from sklearn.impute import SimpleImputer
imputer = SimpleImputer(missing_values=np.nan, strategy='mean')

imputer.fit(X[:, 2:3]) # indexes acc to X
X[:,2:3] = imputer.transform(X[:,2:3])
```

```python
print(X[:,2]) ## Mising values of Age is replaced by mean of all ages
```

### Embarked : text/categorical data, so we apply most_frequent startegy

```python
print(X[:,6])   ## checking missing values for Embarked
```

```python
imputer_1= SimpleImputer(missing_values=np.nan , strategy='most_frequent')
imputer_1.fit(X[:,6:7]) # indexes acc to X
X[:, 6:7] = imputer_1.transform(X[:,6:7])
```

```python
print(X[:,6]) ## Mising values of Embarked is replaced by most frequents of Embarked
```

## Encoding categorical data

### Encoding the Independent Variable

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

ct = ColumnTransformer(transformers=[('encoder', OneHotEncoder(), [1,6])], remainder='passthrough')
X= np.array(ct.fit_transform(X))
```

```python
print(ct.get_feature_names_out()) ##encoded variables will come first, and then remaining.... 
```

### Encoding the Dependent Variable

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
y = np.array(le.fit_transform(y))
```

```python
print(y) ## label enoded 
```

## Splitting the dataset into the Training set and Test set

```python
print(X[0])
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=0.2, random_state=1)
```

```python
print(pd.DataFrame(X_train))
```

```python
print(X_test)
```

```python
print(y_train)
```

```python
print(y_test)
```

## Feature Scaling

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
X_train[:,5:] = sc.fit_transform(X_train[:,5:])
X_test[:,5:] = sc.transform(X_test[:,5:])
```

```python
print(pd.DataFrame(X_train))
```

## train model (just for verification)

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
```

```python
model = RandomForestClassifier()
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
```

```python
accuracy = accuracy_score(y_test, y_pred)
print(accuracy)
```
